```python
import os
from onnx2tf import convert

# modelo onnx
onnx_model_path = '/content/best.onnx'

# ruta salida
saved_model_output_dir = '/content/saved_model_tf'
os.makedirs(saved_model_output_dir, exist_ok=True)

print(f"Converting ONNX model from: {onnx_model_path}")
print(f"Saving TensorFlow SavedModel to: {saved_model_output_dir}")

# Convertir ONNX a TensorFlow SavedModel
# Basado en el traceback, la firma es convert(input_onnx_file_path, onnx_graph, output_folder_path, ...)
convert(
    onnx_model_path, # input_onnx_file_path (1er argumento posicional)
    None,            # onnx_graph (2do argumento posicional, None cuando se proporciona una ruta de archivo)
    saved_model_output_dir # output_folder_path (3er argumento posicional)
)

print("Conversion to TensorFlow SavedModel completed successfully.")
```

```python
import tensorflow as tf
import os

# Define la ruta al SavedModel de TensorFlow (del paso anterior)
saved_model_output_dir = '/content/saved_model_tf'

# Define la ruta de salida para el modelo TFLite
tflite_model_path = '/content/best_quantized.tflite'

print(f"Cargando SavedModel desde: {saved_model_output_dir}")

# Crea un objeto TFLiteConverter a partir del SavedModel
converter = tf.lite.TFLiteConverter.from_saved_model(saved_model_output_dir)

# Habilita optimizaciones, incluyendo la cuantificación de rango dinámico
converter.optimizations = [tf.lite.Optimize.DEFAULT]

# Convierte el modelo a formato TFLite
tflite_model = converter.convert()

# Guarda el modelo TFLite en un archivo
with open(tflite_model_path, 'wb') as f:
    f.write(tflite_model)

print(f"Conversión a TFLite completada exitosamente. Modelo guardado en: {tflite_model_path}")
```
